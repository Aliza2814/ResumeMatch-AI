# 01. Data Exploration & Discovery (ResumeMatch AI)

This notebook inspects the raw datasets for ResumeMatch AI:
- `resumes.csv` (Updated Resume Dataset)
- `placement_salary.csv` (Placement & Salary Benchmark)
- `job_postings.csv` (Job Postings Catalog)

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

data_dir = Path("../data/raw")
df_resumes = pd.read_csv(data_dir / "resumes.csv")
print("Resume Dataset Shape:", df_resumes.shape)
df_resumes.head()
```

```python
# Category Distribution in Resumes
plt.figure(figsize=(12, 6))
df_resumes['Category'].value_counts().plot(kind='bar', color='#3B82F6')
plt.title("Distribution of Resume Categories (25 Domains)")
plt.xlabel("Domain Category")
plt.ylabel("Number of Resumes")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Inspect Placement & Salary Dataset
df_placement = pd.read_csv(data_dir / "placement_salary.csv")
print("Placement Data Shape:", df_placement.shape)
df_placement.describe()
```
